```python
import random
from collections import Counter
import statistics

def generate_uniform_list(a: int, b: int, length: int) -> list:
    """
    Возвращает список длины length из равномерно распределённых
    случайных целых чисел в диапазоне [a, b] включительно.
    """
    if length <= 0:
        return []
    return [random.randint(a, b) for _ in range(length)]
```

```python
def frequency_dict(lst: list) -> dict:
    """
    Принимает список и возвращает словарь {значение: количество вхождений}.
    Пример: [1, 2, 3, 1, 2, 2] -> {1: 2, 2: 3, 3: 1}
    """
    return dict(Counter(lst))
```

```python
def list_statistics(lst: list) -> dict:
    """
    Возвращает словарь с минимумом, средним, медианой, модой и максимумом.
    """
    if not lst:
        return {}
    return {
        'min': min(lst),
        'mean': statistics.mean(lst),
        'median': statistics.median(lst),
        'mode': statistics.mode(lst),
        'max': max(lst)
    }
```

```python
# Пример использования
a, b = 1, 10
length = 20

random_list = generate_uniform_list(a, b, length)
print("Сгенерированный список:", random_list)

freq = frequency_dict(random_list)
print("Частотный словарь:", freq)

stats = list_statistics(random_list)
print("Статистики:", stats)
```

```python
print(f"Список: {random_list}")
print(f"Уникальные значения и их частоты: {freq}")
print(f"Минимум: {stats['min']}")
print(f"Среднее: {stats['mean']:.2f}")
print(f"Медиана: {stats['median']}")
print(f"Мода: {stats['mode']}")
print(f"Максимум: {stats['max']}")
```
